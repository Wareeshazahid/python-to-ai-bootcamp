```python
#Getting the data
import pandas as pd
import numpy as np

df = pd.read_csv("titanic.csv")

df.head()
```

```python
#Looking at the data
print(df.shape)

print(df.columns)

df.info()
```

```python
#Checking missing values
print(df.isnull().sum())

print(df.describe())
```

```python
#Checking survival
print(df["Survived"].value_counts())

print(df["Survived"].mean())

print(df.groupby("Sex")["Survived"].mean())
```

```python
#Checking passenger details
print(df["Pclass"].value_counts())

first_class = df[df["Pclass"] == 1]
print(first_class.head())

adults = df[df["Age"] >= 18]
print(len(adults))
```
