```python
#Creating NumPy Arrays
import numpy as np

# creating arrays
marks = np.array([78, 85, 42, 60, 90])
print(marks)
```

```python
#Checking Array Information
print("Array:", marks)
print("Number of elements:", marks.size)
print("Dimensions:", marks.ndim)
print("Shape:", marks.shape)
print("Data type:", marks.dtype)
```

```python
#Accessing Array Elements
print("First mark:", marks[0])
print("Third mark:", marks[2])
print("Last mark:", marks[-1])
```

```python
#Selecting Marks from an Array
print("First three marks:", marks[:3])
print("Marks from second to fourth:", marks[1:4])
```

```python
# Array Calculations
print("Marks after adding 5:", marks + 5)
print("Marks after subtracting 5:", marks - 5)
print("Marks doubled:", marks * 2)
```

```python
#Finding Average and Highest Marks
print("Total marks:", np.sum(marks))
print("Average marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
```

```python
#Two-Dimensional Arrays
students = np.array([
    [78, 85, 90],
    [65, 70, 75],
    [88, 92, 80]
])

print(students)
print("First student's marks:", students[0])
print("Second student's first mark:", students[1, 0])
```

```python
#Calculating Student Averages
students = np.array([
    [78, 85, 90],
    [65, 70, 75],
    [88, 92, 80]
])

averages = np.mean(students, axis=1)

print("Student averages:", averages)
```

```python
#Finding Marks Above 75
high_marks = marks[marks > 75]

print("Marks above 75:", high_marks)
```
