```python
#A small practice project
name = input("Enter student name: ")
marks = float(input("Enter marks out of 100: "))

if marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print("Student:", name)
print("Marks:", marks)
print("Grade:", grade)
```

```python
# Student Marks Analyzer
def analyze_marks(marks):
    total = sum(marks)
    average = total / len(marks)

    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"

    print("Marks:", marks)
    print("Total:", total)
    print("Average:", round(average, 2))
    print("Result:", result)


student_marks = [78, 85, 42, 6, 90]
analyze_marks(student_marks)
```
